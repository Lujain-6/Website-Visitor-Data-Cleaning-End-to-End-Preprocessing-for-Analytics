```python
# This code is used for outlier detection and removal based on the Z-score method. It filters rows in the raw_data DataFrame that contain outliers in the columns specified by the out_list (i.e., value, monthly visits, and item purchased).
# This calculates the Z-score for each value in the columns specified in out_list. The Z-score measures how many standard deviations a data point is from the mean of its respective column.
# The Z-score is calculated as  z= (x-mu)/sigma
# Z-scores near 0 indicate values close to the mean, while larger absolute Z-scores indicate potential outliers.
# This condition filters the rows where the absolute Z-score is less than 3. A Z-score greater than 3 (or less than -3) typically indicates an outlier, as it means the data point is more than 3 standard deviations away from the mean.
# So, this filters out values that are within 3 standard deviations of the mean, retaining "normal" data points and excluding potential outliers.
# .all(axis=1): This ensures that the condition (absolute Z-score < 3) holds for all columns in out_list simultaneously. In other words, it keeps rows where all values in the selected columns have Z-scores below 3.
# If any column in a row has a Z-score greater than or equal to 3 (indicating a potential outlier), the entire row is excluded from raw_data.

out_list = ['value','monthly visits', 'item purchased']

raw_data = raw_data[(np.abs(stats.zscore(raw_data[out_list]))<3).all(axis=1)]
```
